```python
# Download the Titanic dataset directly
import pandas as pd

# Load dataset from the internet
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully! ✅")
```

```python
# See the first 5 rows
df.head()
```

```python
# Get information about the dataset
df.info()
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()
print(missing_values)
```

```python
# Fill missing ages with the average age
average_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(average_age)

print("Missing ages fixed! ✅")
```

```python
# Remove the Cabin column (too many missing values)
df = df.drop('Cabin', axis=1)

print("Removed Cabin column! ✅")
```

```python
# Check how many rows before
print(f"Rows before: {len(df)}")

# Remove duplicates
df = df.drop_duplicates()

# Check how many rows after
print(f"Rows after: {len(df)}")
```

```python
# Create age categories
def categorize_age(age):
    if age < 18:
        return 'Child'
    elif age < 60:
        return 'Adult'
    else:
        return 'Senior'

df['Age_Group'] = df['Age'].apply(categorize_age)

print("New Age_Group column created! ✅")
```

```python
# Save to a new CSV file
df.to_csv('cleaned_titanic_data.csv', index=False)

print("Cleaned data saved! ✅")
```

```python
# Import library
import pandas as pd

# Load dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Step 1: Explore data
print("First 5 rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

# Step 2: Find missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 3: Clean missing values
# Fill missing ages with average
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Fill missing Embarked with most common value
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Drop Cabin column (too many missing)
df.drop('Cabin', axis=1, inplace=True)

# Step 4: Remove duplicates
print(f"\nRows before removing duplicates: {len(df)}")
df.drop_duplicates(inplace=True)
print(f"Rows after removing duplicates: {len(df)}")

# Step 5: Create new column
def categorize_age(age):
    if age < 18:
        return 'Child'
    elif age < 60:
        return 'Adult'
    else:
        return 'Senior'

df['Age_Group'] = df['Age'].apply(categorize_age)

# Step 6: Save cleaned data
df.to_csv('cleaned_titanic_data.csv', index=False)

print("\n✅ Data cleaning completed!")
print(f"Final dataset shape: {df.shape}")
```
